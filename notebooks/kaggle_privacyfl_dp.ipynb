```python
from dataclasses import dataclass
from hashlib import sha256

@dataclass(frozen=True)
class SyntheticClient:
    client_id: int
    record_ids: tuple[str, ...]

clients = tuple(SyntheticClient(i, tuple(f'client{i}_record{j}' for j in range(3))) for i in range(5))
all_ids = [rid for client in clients for rid in client.record_ids]
assert len(clients) == 5
assert len(all_ids) == len(set(all_ids)), 'synthetic client overlap detected'

labels = ['0.1', '0.5', '1.0', '5.0', '10.0', 'inf']
records = []
for label in labels:
    target_epsilon = None if label == 'inf' else float(label)
    records.append({
        'label': label,
        'target_epsilon': target_epsilon,
        'target_delta': None if target_epsilon is None else 1e-5,
        'max_grad_norm': None if target_epsilon is None else 1.0,
        'total_epochs': 10 * 3,
        'status': 'synthetic_contract_only'
    })

payload = json.dumps(records, sort_keys=True).encode('utf-8')
print('synthetic_clients=', len(clients))
print('operating_labels=', labels)
print('contract_sha256=', sha256(payload).hexdigest())
```
